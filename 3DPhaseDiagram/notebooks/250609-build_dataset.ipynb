# Load Initial Data


```python
df1 = pd.read_csv('figure2.csv') #actually has both subplots
phase_names = df1.columns[::2].values
column_index = pd.MultiIndex.from_arrays([
    np.repeat(phase_names, 2),  # Repeat each phase name twice
    ['X', 'Y'] * len(phase_names)  # Alternate X and Y
])
df1.columns = column_index
df1 = df1.iloc[1:].reset_index(drop=True)
df1
```

build the multiindex and add in the extra column for the first slice (with 'v2' in the name)

```python
# Get columns with 'v2' in their names for potential future use (currently not used)
v2_columns = [col for col in df1.columns if 'v2' in col[0]]

# Create a dictionary to hold our new multiindex columns data
new_data = {}

# Process each phase to build columns
for phase in ['phase1v2', 'phase2v2', 'phase3v2']:
    # Define the original X and Y column keys for the phase
    x_col = (phase, 'X')
    y_col = (phase, 'Y')
    
    # Add the renamed columns to our dictionary
    new_data[(phase, 'protein')] = df1[x_col]
    new_data[(phase, 'temperature')] = df1[y_col]
    new_data[(phase, 'glycerol')] = 0

# Create the new dataframe from the dictionary
new_df = pd.DataFrame(new_data)

# Convert the columns to a proper MultiIndex with descriptive names
new_df.columns = pd.MultiIndex.from_tuples(new_df.columns, names=['phase', 'measurement'])

# Display the new dataframe
new_df
```

do the same for the second slice

```python
# Get columns without 'v2' in their names
not_v2_columns = [col for col in df1.columns if 'v2' not in col[0]]

# Create a dictionary to hold our new multiindex columns data
new_data_not_v2 = {}

# Process each phase to build columns
for phase in ['phase1', 'phase2', 'phase3']:
    # Define the original X and Y column keys for the phase
    x_col = (phase, 'X')
    y_col = (phase, 'Y')
    
    # Add the renamed columns to our dictionary
    new_data_not_v2[(phase, 'glycerol')] = df1[x_col]
    new_data_not_v2[(phase, 'temperature')] = df1[y_col]
    new_data_not_v2[(phase, 'protein')] = 50

# Create the new dataframe from the dictionary
new_df_not_v2 = pd.DataFrame(new_data_not_v2)

# Convert the columns to a proper MultiIndex with descriptive names
new_df_not_v2.columns = pd.MultiIndex.from_tuples(new_df_not_v2.columns, names=['phase', 'measurement'])

# Display the new dataframe
new_df_not_v2
```

combine the common 'phases' from each slice into single dataframes

```python
df_list = []
for phasename in ['phase1', 'phase2', 'phase3']:
    # Drop NaN values from 'phase1' and 'phase1v2' in new_df
    df1 = new_df[phasename+'v2'].dropna().astype(float)
    df2 = new_df_not_v2[phasename].dropna().astype(float)

    # Combine 'phase1' and 'phase1v2' in new_df_not_v2
    df12 = pd.concat([df1, df2], axis=0, ignore_index=True)
    display(df12)

    df_list.append(df12)
```

```python
import plotly.express as px
import plotly.graph_objects as go

# Create a 3D scatter plot for each dataframe in df_list
fig = go.Figure()

# Define colors for each phase
colors = ['red', 'green', 'blue']

for i, df in enumerate(df_list):
    fig.add_trace(go.Scatter3d(
        x=df['protein'],
        y=df['temperature'],
        z=df['glycerol'],
        mode='markers',
        marker=dict(size=5, color=colors[i], opacity=0.8),
        name=f'Phase {i+1}'
    ))

# Update layout for better visualization
fig.update_layout(
    scene=dict(
        xaxis_title='Protein Concentration',
        yaxis_title='Temperature',
        zaxis_title='Glycerol Concentration'
    ),
    title='3D Scatter Plot of Phases',
    margin=dict(l=0, r=0, b=0, t=50)
)

# Show the plot
fig.show()
```

finally, make master dataset

```python
import xarray as xr

ds_list = []
for i, df in enumerate(df_list):


    ds = xr.Dataset()
    ds['composition'] = df.to_xarray().to_array('component')
    ds['labels'] = ('index',np.ones(ds.sizes['index'])*i)
    ds = ds.rename({'index': 'sample'})
    ds_list.append(ds)

ds = xr.concat(ds_list, dim='sample')
ds = ds.transpose('sample', 'component')

ds
```

Now we (extap/interp)polate over the entire composition space

```python
with Pipeline(name = "find_boundaries") as p:

    CartesianGrid(
        output_variable="composition_grid",
        sample_dim="grid",
        grid_spec = {'protein':{'min': 0.0, 'max': 80.0, 'steps': 25},
                     'glycerol':{'min': 0.0, 'max': 12.0, 'steps': 25},
                     'temperature':{'min': 0.0, 'max': 50.0, 'steps': 25}}
    )

    Standardize(
        input_variable="composition",
        output_variable="normalized_composition",
        dim="sample",
        component_dim="component",
        scale_variable=None,
        min_val={'protein': 0.0, 'glycerol': 0.0, 'temperature': 0.0},
        max_val={'protein': 80.0, 'glycerol': 12.0, 'temperature': 50.0},
        name="Standardize",
    )

    Standardize(
        input_variable="composition_grid",
        output_variable="normalized_composition_grid",
        dim="grid",
        component_dim="component",
        scale_variable=None,
        min_val={'protein': 0.0, 'glycerol': 0.0, 'temperature': 0.0},
        max_val={'protein': 80.0, 'glycerol': 12.0, 'temperature': 50.0},
        name="Standardize",
    )

    GaussianProcessClassifier(
        feature_input_variable="normalized_composition",
        predictor_input_variable="labels",
        output_prefix="extrap",
        grid_variable="normalized_composition_grid",
        grid_dim="grid",
        sample_dim="sample",
        kernel="Matern",
        kernel_kwargs={'length_scale': 1.0, 'nu': 1.5},
        optimizer="fmin_l_bfgs_b",
        name="GaussianProcessClassifier",
    )

p.print()
```

# Expand Dataset 

One last step, we need an 'expanded' version of each phase. This is so that compositions near the boundaries still work. We'll expand each part of the grid around its center of mass

```python
# Set expansion percentage (e.g., 10% expansion)
expansion_percentage = 0.1
boundary_dataset2 = boundary_dataset.copy()

# Group by labels and expand positions
expanded_compositions = []
expanded_labels= []
for label, sds in boundary_dataset2.groupby('labels'):
    composition = sds['composition']
    center = composition.mean('sample')
    vectors_from_center = composition - center
    # Expand each point radially from the center by the expansion percentage
    expanded_composition = center + (1 + expansion_percentage) * vectors_from_center
    expanded_compositions.append(expanded_composition)
    expanded_labels.append(sds.labels)

expanded_composition = xr.concat(expanded_compositions,dim='sample')
expanded_labels = xr.concat(expanded_labels,dim='sample')

boundary_dataset2['composition'] = expanded_composition
boundary_dataset2['labels'] = expanded_labels
boundary_dataset2
```

```python
plot_scatter_plotly(
    dataset=boundary_dataset2,
    component_variable='composition',
    labels='labels',
    discrete_labels=True,
)
```
